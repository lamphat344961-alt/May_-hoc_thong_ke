```python
import pandas as pd 
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
df= pd.read_excel("data (1)\\data\\Sales.xlsx")
x= df[['Price','Ads_Cost']]
y= df[['Sales_Volume']]
x,y = np.array(x), np.array(y)
x =sm.add_constant(x)
```

```python
# Creta model 
model= sm.OLS(y,x)
results =model.fit()
```

```python
R_sq=results.rsquared
params = results.params
print(results.summary())
```
