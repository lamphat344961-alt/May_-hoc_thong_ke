```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm  
import pickle
import warnings
warnings.filterwarnings('ignore')
```

```python
plt.rcParams['figure.figsize'] = (10,8)
plt.rcParams['figure.dpi'] = 100 
plt.rcParams['font.size'] = 14 
```

```python
df = pd.read_csv('..\data (2)\data\Income.csv')
```

```python
Income = df['Income']
Expenditure = df['Expenditure']
```

```python
Income , Expenditure = np.array(Income), np.array(Expenditure)
```

```python
Income = sm.add_constant(Income)
```

```python
model = sm.OLS(Expenditure, Income)
```

```python
results = model.fit()
```

```python
R_sq = results.rsquared
params = results.params
print(results.summary())
```

```python
plt.scatter(df['Income'], df['Expenditure'])
plt.plot(df['Income'], params[0] + params[1] * df['Income'])
plt.show()
```

```python
predict_values = results.predict(Income)
```

```python
plt.scatter(df['Income'], df['Expenditure'], label = 'Actual')
plt.plot(df['Income'], predict_values, color = 'r', label = 'Predict', linestyle = '--', marker = 'x')
plt.xlabel('Income')
plt.ylabel('Expenditure')
plt.legend()
plt.show()
```
