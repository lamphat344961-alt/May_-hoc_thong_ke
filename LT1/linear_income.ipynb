```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import  LinearRegression
import pickle
import warnings
warnings.filterwarnings('ignore')
```

```python
plt.rcParams['figure.figsize'] = (10,8)
plt.rcParams['figure.dpi'] = 100 
plt.rcParams['font.size'] = 14 
```

```python
df = pd.read_csv('data (2)\data\Income.csv')
```

```python
df
```

```python
x = df[['Income']]
y = df[['Expenditure']]
```

```python
model = LinearRegression().fit(x, y)
```

```python
intercept = model.intercept_
slope = model.coef_
R_2 = model.score(x, y)
```

```python
intercept
```

```python
slope
```

```python
plt.scatter(x,y)
plt.plot(x, intercept + slope * x)
plt.xlabel('Income')
plt.ylabel('Expenditure')
plt.show()
```

```python
pre_value = model.predict(x)
plt.plot(x, y, label='Actual', color = 'red', linestyle = '--')
plt.plot(x, pre_value , label = 'predict')
plt.legend()
plt.show()
```

```python
others_incomes = np.array([35, 42, 28]).reshape(-1,1)
pre_future_value = model.predict(others_incomes)
```
