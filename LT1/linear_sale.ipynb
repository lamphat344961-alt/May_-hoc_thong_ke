```python
import pandas as pd 
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import pickle
```

```python
df= pd.read_excel("data (1)\\data\\Sales.xlsx")
```

```python
df.head()
```

```python
x = df[['Price','Ads_Cost']]
y = df[['Sales_Volume']]
model = LinearRegression().fit(x, y)
```

```python
R_sq = model.score(x, y)
intercept = model.intercept_
coefs = model.coef_

```

predict for present values 


```python
predicted_values = model.predict(x)
plt.plot(y, label="Actual")
plt. plot(predicted_values, color='darkblue', label="Predicted", linestyle='--', marker='x')
plt.legend()
plt.show()
```

predict for future values 

```python
x_new= [[4.2,4.8],[4.8,4.3]]
predicted_future_values = model.predict(x_new)
```

# split train/ test dataset 
x_train, x_test, y_train

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.1)
```

```python
new_model= LinearRegression().fit(x_train,y_train)
score= new_model.score(x_train, y_train)
print("R2:", score)
print(y_test)
y_pred = new_model.predict(x_test)
print(y_pred[0][0])
```

```python
with open('model/linear_sale.pkl', 'wb') as f :
    pickle.dump(new_model, f)
```
