```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams['figure.figsize'] = (10,8)
plt.rcParams['figure.dpi'] = 100 
plt.rcParams['font.size'] = 14 
```

```python
def cost_function (theta) : 
    return 12*theta**3 - 3*theta**2 + 10
```

```python
def gradient (theta) : 
    return 36*theta**2 - 6*theta
```

```python
def gradient_descent (starting_theta, learning_rate, eps) : 
    theta = starting_theta
    theta_history = [theta]
    cost_history = [cost_function(theta)]

    while True : 
        theta = theta - learning_rate * gradient(theta)
        theta_history.append(theta)
        cost_history.append(cost_function(theta))
        if abs(gradient(theta)) < eps : 
            break
    return theta_history, cost_history
```

```python
starting_theta = 3
learning_rate = 0.00001
eps =0.0001 
```

```python
theta_history, cost_history = gradient_descent(starting_theta, learning_rate, eps)
```

```python
theta_values = np.linspace(-5, 5, 100)
cost_values = cost_function(theta_values)
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(theta_values, cost_values, label='&J(\\theta) = \\theta^4 - 6\\theta^2 - 2\\theta + 10$', color = "darkgreen", lw = 2)
plt.scatter(theta_history, cost_history, color='darkorange', label='Gradient Descent Steps', s= 88 )
plt.plot(theta_history, cost_history, color='darkorange', lw = 2, linestyle='--')
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(theta_values, cost_values, label='&J(\\theta) = 12\\theta - 3\\theta^2 + 10$', color = "darkgreen", lw = 2)
plt.scatter(theta_history, cost_history, color='red', label='Gradient Descent Steps', s= 80 )
plt.plot(theta_history, cost_history, color='red', lw = 2)
```

```python
plt.subplot(1, 2, 2)
plt.plot(cost_history, label = 'Cost Function', color='darkgreen', lw = 2) 
plt.xlabel('Iteration')
plt.ylabel('Cost - $J(\\theta)$')
plt.title('Cost Function Value over Iterations')
plt.legend()
plt.tight_layout()
plt.show()
```
