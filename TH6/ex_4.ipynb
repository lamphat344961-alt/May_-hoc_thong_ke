```python

import numpy as np 
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

import matplotlib.pyplot as plt
```

```python
plt.rcParams['figure.figsize'] = (10,8)
plt.rcParams['figure.dpi'] = 100 
plt.rcParams['font.size'] = 14 
```

```python
X = np.array([i for i in range (16)]).reshape(-1,1)
noise_std_dev = 0.5 
np.random.seed(40)
noise = np.random.normal(loc=0 , scale = noise_std_dev , size = X.shape )
y= X + noise 

X_b = np.c_[np.ones((len(X), 1 )), X] 
print(X_b.shape[1])
```

```python
plt.scatter(X, y , color = 'darkgreen', s= 80) 
plt.xlabel ('X', fontweight = 'bold')
plt.ylabel ('y', fontweight = 'bold')
plt.show()
```

```python
model = Sequential ()
model.add(keras.Input(shape=(X_b.shape[1],)))
model.add(Dense(units=1, activation='linear'))
```

```python
model.compile(optimizer='adam', loss = 'mean_squared_error')
```

```python
history = model.fit(X_b, y, epochs=2000) 
```

```python
plt.plot(history.history['loss'], label = 'Train Loss', color = 'darkgreen', lw= 3 )
plt.xlabel('epochs')
plt.ylabel('loss')
plt.legend()
plt.show()
```

```python
loss = model.evaluate(X_b, y)
print(f'training loss : {loss}')
```

```python
y_pred = model.predict(X_b)
```

```python
plt.scatter(X, y, label = "True" , color = 'darkgreen', s = 80) 
plt.plot(X, y_pred, color ='darkorange', label = 'Pre')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()

```
